```python
from transformers import pipeline, GPT2Tokenizer, GPT2Model, GPT2LMHeadModel, BertTokenizer, BertForSequenceClassification
import matplotlib.pyplot as plt
import seaborn as sns
import torch
import numpy as np
```

##Generate Article Using GPT-2

```python
generator = pipeline("text-generation", model="gpt2")

prompt = "Climate change and renewable energy investments in 2025"
# Use max_length to control the length of the generated text
result = generator(prompt, max_length=100, num_return_sequences=1, truncation=True)

generated_text = result[0]['generated_text']
print("📝 GPT-Generated Report:\n", generated_text)
```

##Evaluate Readability with Perplexity Score

```python
gpt_model = GPT2LMHeadModel.from_pretrained("gpt2")
tokenizer = GPT2Tokenizer.from_pretrained("gpt2")

input_ids = tokenizer.encode(generated_text, return_tensors="pt")
with torch.no_grad():
    loss = gpt_model(input_ids, labels=input_ids).loss
    perplexity = torch.exp(loss)

print(f"📊 Perplexity of Generated Text: {perplexity.item():.2f}")
```

##Run Sentiment Analysis using BERT

```python
classifier = pipeline("sentiment-analysis", model="distilbert/distilbert-base-uncased-finetuned-sst-2-english")

text_for_bert = generated_text[:512]
sentiment = classifier(text_for_bert)

print("🧾 Sentiment Result:\n", sentiment)
```

##Visualize GPT-2 Attention (Token-Level)

```python
# Add attn_implementation="eager" to remove the warning
model = GPT2Model.from_pretrained("gpt2", output_attentions=True, attn_implementation="eager")
tokenizer = GPT2Tokenizer.from_pretrained("gpt2")

prompt = "Climate change and renewable energy investments in 2025" # Use the same prompt as before
inputs = tokenizer(prompt, return_tensors="pt")
outputs = model(**inputs)

# Visualize attention of the last layer
attention = outputs.attentions[-1][0][0].detach().numpy()
tokens = tokenizer.convert_ids_to_tokens(inputs["input_ids"][0])

# Configure matplotlib to use a font that supports the glyph
plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['font.sans-serif'] = ['DejaVu Sans', 'Segoe UI', 'Liberation Sans', 'Roboto', 'Helvetica', 'Arial']


sns.heatmap(attention, xticklabels=tokens, yticklabels=tokens, cmap="YlGnBu")
plt.title("🎯 GPT-2 Self-Attention Heatmap")
plt.xticks(rotation=90)
plt.yticks(rotation=0)
plt.show()
```
